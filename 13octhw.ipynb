```python
# l = [5,4,5,4,6,7,8,4,3,5,3,4]
# WAP to get unique elements from list without changing data's order or position 
# unique_list = [5,4,6,7,8,3]

l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]

unique_list = []

for item in l:
    if item not in unique_list:
        unique_list.append(item)

print(unique_list)
```

```python
# Find frequency of every element from above input list in dict
# freq_list = {5:3, 4:4 , 6:1 , 7:1 , 8:1 , 3:2 }

l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]

freq_list = {}

for item in l:
    if item in freq_list:
        freq_list[item] = freq_list[item] + 1
    else:
        freq_list[item] = 1

print(freq_list)
```

```python
# s = " I love python programming "
# Find frequency of each char n also count of spaces.

s = " I love python programming "

char_freq = {}

for char in s:
    if char in char_freq:
        char_freq[char] = char_freq[char] + 1
    else:
        char_freq[char] = 1

for char in char_freq:
    if char == " ":
        print("space:", char_freq[char])

print(char_freq)
    
```

```python
# db = { "name" : ["jay","raj","ria"]},"marks" : [[78,67,"AB "],[98,"AB",81],["AB","AB",99]],"sub":["maths","chem","phy"]
# WAP to print which std was absent for which sub test

db = {
    "name": ["jay", "raj", "ria"],
    "marks": [
        [78, 67, "AB "],
        [98, "AB", 81],
        ["AB", "AB", 99]
    ],
    "sub": ["maths", "python", "english"]
}

for i in range(len(db["name"])):
    absent_subjects = []

    for j in range(len(db["sub"])):
        mark = db["marks"][i][j]

        if isinstance(mark, str) and mark.strip() == "AB":
            absent_subjects.append(db["sub"][j])

    print(db["name"][i], "absent for : ", absent_subjects)
```
